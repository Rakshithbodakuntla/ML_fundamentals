```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Titanic-Dataset.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.Age.fillna(df.Age.mean(), inplace=True)
```

```python
df['Cabin'].fillna(df['Cabin'].mode()[0], inplace=True)
```

```python
df.Embarked.fillna(df.Embarked.mode()[0], inplace =True)
```

```python
df.head()
```

```python
sns.histplot(x='Age', data=df)
plt.title('Age Distribution')
plt.show()
```

```python
sns.countplot(x='Pclass', hue='Survived', data=df)
plt.title('Pclass')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(x='Sex', hue='Survived', data=df)
plt.title('Survival Count by Gender')
plt.xlabel('Sex')
plt.ylabel('Number of Passengers')
plt.legend(title='Survived', loc='upper right', labels=['No', 'Yes'])
plt.show()
```

```python
df['Cabin'] = df['Cabin'].astype(str).str[0]
display(df[['Cabin']])
```

```python
plt.figure(figsize=(10, 6))
sns.countplot(x='Cabin', hue='Survived', data=df)
plt.title('Survival Count by Cabin Deck')
plt.xlabel('Cabin Deck')
plt.ylabel('Number of Passengers')
plt.show()
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
```

```python
df['Sex'] = le.fit_transform(df['Sex'])
df['Embarked'] = le.fit_transform(df['Embarked'])
df['Cabin'] = le.fit_transform(df['Cabin'])
df.head()
```

```python
x = df[['Pclass','Sex','Age']]
y = df['Survived']
```

```python
correlation_matrix = df[['Pclass', 'Sex', 'Age', 'Survived']].corr()
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True)
plt.title('Correlation Matrix of Pclass, Sex, Age, and Survived')
plt.show()
```
